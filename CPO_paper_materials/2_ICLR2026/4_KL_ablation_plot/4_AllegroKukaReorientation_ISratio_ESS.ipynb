```python

def get_IS_deviation(csv_path, target_step, time_window=time_window):
    csv_data = pd.read_csv(csv_path)
    # データをNumPy配列に変換
    data_numpy = csv_data.to_numpy()
        
    # ターゲットステップに最も近い行を見つける
    idx = int(np.abs(target_step / horizon_length / num_envs))
    idxs = range(max(0, idx - (time_window-1)//2), min(len(data_numpy), idx + (time_window-1)//2))
    #data = data_numpy[idx]
    # 複数行の平均を取る
    if time_window == 1:
        return data_numpy[idx]
    else:
        data = data_numpy[list(idxs)].mean(axis=0)
    
    return data

def get_overall_ess(csv_path, target_step, time_window=time_window):
    csv_data = pd.read_csv(csv_path)
    # データをNumPy配列に変換
    data_numpy = csv_data.to_numpy()
        
    # ターゲットステップに最も近い行を見つける
    idx = int(np.abs(target_step / horizon_length / num_envs))
    idxs = range(max(0, idx - (time_window-1)//2), min(len(data_numpy), idx + (time_window-1)//2))
    #data = data_numpy[idx]
    # 複数行の平均を取る
    if time_window == 1:
        return data_numpy[idx]  # overall_essは2列目にある
    else:
        return data_numpy[list(idxs)].mean()  # overall_essは2列目にある

print("ShadowHand")
print("horizon_length:", horizon_length)
print("num_envs:", num_envs)
print("time_window:", time_window)

print("# --- Mean Followers' IS ratio deviation --- #")
for key in data_configs.keys():     
    csv_path = f"{data_configs[key]['run_dir']}/IS_ratio_deviation_mean.csv"
    for target_step in target_steps:
        
        is_deviation = get_IS_deviation(csv_path, target_step, time_window=time_window)
        followers_deviation = is_deviation[:-1].mean()  # フォロワーのIS ratioの平均
        print(f"{key} at {target_step}: {followers_deviation}")

print("# --- Overall ESS Rate --- #")
for key in data_configs.keys():     
    csv_path = f"{data_configs[key]['run_dir']}/overall_ess.csv"
    for target_step in target_steps:
        overall_ess = get_overall_ess(csv_path, target_step, time_window=time_window)
        print(f"{key} at {target_step}: {overall_ess}")
```
