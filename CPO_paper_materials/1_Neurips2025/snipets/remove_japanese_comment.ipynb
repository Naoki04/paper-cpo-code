```python
import os
import re

def remove_japanese_comments_and_docstrings(text):
    # 日本語を含む行内コメント（#〜）
    text = re.sub(r'[#＃].*[\u3040-\u30FF\u4E00-\u9FFF].*', '', text)

    # 日本語を含む docstring（"""〜""" or '''〜'''）
    pattern = r'(?:[rRuU]{0,2})("""|\'\'\')(.*?[\u3040-\u30FF\u4E00-\u9FFF].*?)(\1)'
    text = re.sub(pattern, '', text, flags=re.DOTALL)

    return text

def process_file(path):
    with open(path, 'r', encoding='utf-8') as f:
        content = f.read()
    new_content = remove_japanese_comments_and_docstrings(content)
    with open(path, 'w', encoding='utf-8') as f:
        f.write(new_content)


process_file("/home/mil/shitanda/CPO_NeurIPS2025/supplementary_code/rl_games/rl_games/common/a2c_common.py")
#for root, _, files in os.walk(target_dir):
    #for file in files:
     #   if file.endswith(".py"):
      #      process_file(os.path.join(root, file))
```
