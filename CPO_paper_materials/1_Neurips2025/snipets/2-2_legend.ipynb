```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# =======================
# 凡例に表示する手法と設定
# =======================
colours = {
    "CPO (w/AdR)": "red",      # 提案手法（敵対的報酬あり）
    "CPO (wo/AdR)": "blue",    # 提案手法（敵対的報酬なし）
}

styles = {
    "CPO (w/AdR)": "-",         # 実線
    "CPO (wo/AdR)": "--",       # 破線
}

# =======================
# 凡例用の Line2D オブジェクト作成（線の長さを長く）
# =======================
legend_elements = []
for label in colours:
    line = Line2D(
        [0, 2], [0, 0],  # ← 線の長さを長くする
        color=colours[label],
        linestyle=styles[label],
        linewidth=2.5,
        label=label
    )
    legend_elements.append(line)

# =======================
# 凡例だけの縦長図を作成
# =======================
fig, ax = plt.subplots(figsize=(2.6, 1.8))  # 少し横幅増加
ax.axis("off")

# 凡例を上部中央に配置（線を長めに表示）
ax.legend(
    handles=legend_elements,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.0),
    frameon=False,
    fontsize=11,
    handlelength=4.0,      # ← 線をさらに長く表示
    borderaxespad=0.0,
)

# 保存（余白最小）
fig.savefig("legend_vertical_longline.svg", format="svg", bbox_inches="tight")
```
