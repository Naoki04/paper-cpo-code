```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import torch
import numpy as np

# カラーマップ設定（vmin, vmax をデータに合わせて）
vmin = 0
vmax = 3
cmap = 'viridis'

# ScalarMappable を作成
norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)
sm = mpl.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])  # ダミーデータ設定（必須）

# 図とカラーバーを作成
fig, ax = plt.subplots(figsize=(4.5, 12))  # 縦長に設定（調整可能）
cbar = fig.colorbar(sm, cax=ax)

# フォントサイズなど調整（任意）
cbar.ax.tick_params(labelsize=20)

# SVGとして保存
fig.savefig("colorbar_only.svg", format="svg", bbox_inches='tight', pad_inches=0.1)
```
